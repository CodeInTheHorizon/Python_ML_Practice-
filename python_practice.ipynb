**Challenge 2:**
- *write a python function that takes a list and returns a new list with unique elements of the list*
- *input=[1,2,3,1,2,4]*
- *output=[1,2,3,4]*

```python
lst=[1,2,3,1,2,4]
```

```python
def unique(li):

  new=[]
  # iterate on the li list
  for i in li:
      #we are adding only unique elements in new list
    if i not in new:
      new.append(i)

  #return new list
  return new


```
