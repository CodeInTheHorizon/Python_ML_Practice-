```python
def calculate_average(values):
    assert values, "Cannot calculate average of an empty list"
    total = sum(values)
    count = len(values)
    return total / count

scores1 = [60, 75, 82, 90, 88]
print("Average of scores1:", calculate_average(scores1))

scores2 = []
print("Average of scores2:", calculate_average(scores2))
```

```python
def average_marks(marks):
    if len(marks) != 0:
        return sum(marks) / len(marks)
    else:
        return 0

marks_list = [5, 10, 15, 20]
print("Average of marks_list:", average_marks(marks_list))
```

```python
def calculate_average(numbers):
    try:
        return sum(numbers) / len(numbers)
    except ZeroDivisionError:
        return "Cannot calculate average of an empty list"

list1 = [70, 85, 90, 88]
print("Average of list1:", calculate_average(list1))

list2 = []
print("Average of list2:", calculate_average(list2))
```
